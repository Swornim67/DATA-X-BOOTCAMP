### **SQL Practice: SELECT, FROM, WHERE, DISTINCT, GROUP BY, HAVING, and ORDER BY (PART 2)**

### **BASIC** **LEVEL**

```sql
-- Display first_name, last_name, and salary for employees earning more than 70,000.
select first_name,last_name,salary from employees 
where salary > 70000;
```

```sql
-- Display emp_id, first_name, department, and job_title for employees in the HR
select emp_id,first_name,department,job_title from employees 
where department='HR';
```

```sql
-- Display first_name, last_name, department, and status for all inactive employees.
select first_name,last_name,department,status from employees 
where status='Inactive';
```

```sql
--Find active employees who live in Mumbai. Display emp_id, first_name, city, and status.
select emp_id,first_name,city,status from employees 
where status='Active' and city='Mumbai';
```

```sql
--Find employees in Engineering, Finance, or Marketing. Display first_name, department, and salary.
select first_name,department,salary from employees
where department in ('Engineering','Finance','Marketing');
```

```sql
--Display first_name and salary for employees earning between 60,000 and 90,000, including both endpoints.
select first_name,salary from employees 
where salary between 60000 and 90000;
```

```sql
--Find employees hired after January 1, 2020. Display first_name, hire_date, and department.
select first_name,hire_date,department from employees
where hire_date > '2020-01-01';
```

```sql
-- Display the five highest-paid employees, showing their names and salaries, from highest salary to lowest.
select first_name,last_name,salary from employees 
order by salary desc 
limit 5;
```

```sql
-- Display each department only once, in alphabetical order.
select distinct department from employees 
order by department;
```

```sql
--Display the three most recently hired employees, showing their names and hire dates, with the newest hire first.
select first_name,last_name,hire_date from employees
order by hire_date desc 
limit 3;
```

### **SQL aggregation**

```sql
--Count the total number of employees.
SELECT count(emp_id) from employees;
```

```sql
--Find the average salary of all employees.
select  avg(salary ) from employees
```

```sql
--Display each department and its total number of employees.
select department,count(*) from employees
group by department;
```

```sql
--Display each department and its average salary, sorted from highest average salary to lowest.
select distinct department,avg(salary)as avg_salary from employees
group by department 
order by avg_salary desc;
```

```sql
--Show only departments that have more than 8 employees.
select department,count(*) as total_emp_count
from employees
group by department
having total_emp_count > 8;
```

```sql
--Find the highest salary in each department.
select department,max(salary) from employees
group by department;
```

```sql
--Count how many employees are Active and how many are Inactive.
select status,count(*) from employees
group by status;
```

```sql
--Find the total salary expenditure for each city.
select city,sum(salary) from employees
group by city;
```

```sql
--Display departments whose average salary is greater than 75,000.
select department,avg(salary)as avg_salary from employees
group by department 
having avg_salary > 75000;
```

```sql
--Find the department with the largest number of employees.
select department,count(*)
from employees
group by department
order by count(*) desc limit 1 

```

### **LIKE and pattern matching.**

```sql
--Find employees whose first_name starts with A.
select * from employees 
where first_name like 'a%';
```

```sql
--Find employees whose first_name ends with the letter a.
select * from employees 
where first_name like '%a';
```

```sql
--Find employees whose first_name contains r.
select * from employees 
where first_name like '%r%';
```

```sql
--Find employees whose last_name starts with S.
select * from employees 
where last_name like 's%';
```

```sql
--Find employees whose email address ends with gmail.com.
select * from employees 
where email like '%gmail.com';
```

```sql
--Find employees whose first name contains exactly five characters.
select * from employees
where first_name like '_____';
```

```sql
--Find employees whose first name has a as the second character.
select * from employees 
where first_name like '_a%';
```

```sql
--Find employees whose last name contains sh anywhere.
select * from employees
where last_name like '%sh%';
```

```sql
--Find employees whose first name starts with either A or P.
select * from employees 
where first_name like  'a%' or first_name like 'p%';
```

```sql
--Find employees whose first name does not start with A.
select * from employees 
where first_name not like 'A%';
```

```sql
--Find active employees whose first name starts with A.
select * from employees 
where status='Active' and first_name like 'A%';
```

```sql
--Find employees whose last name ends with a and whose salary is greater than 60,000.
select * from employees 
where last_name like '%a' and salary > 60000;
```

```sql
--Find employees who live in Mumbai or Delhi and whose first name starts with P.
select * from employees
where city in('Mumbai','Delhi') and first_name like 'P%';
```

```sql
--Find employees whose first name contains a but does not start with A.
select * from employees 
where first_name like '%a%'
and first_name not like 'A%';
```

```sql
--Find employees whose email contains gmail.com or yahoo.com.
select * from employees 
where email like '%gmail.com' or email like '%yahoo.com';
```
