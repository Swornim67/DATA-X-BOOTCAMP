### SELECT * VS SELECT NEEDED COLUMN

```python
%sql
-- BADLY OPTIMIZED QUERY (Took 27s)
select * 
from customers c 
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id;  
```

```python
%sql
-- OPTIMIZED QUERY(Took 3s)
select 
c.first_name,c.last_name,od.product_name,od.price_each
from customers c  
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id
```

### FILTERING BEFORE JOIN VS FILTERING AFTER JOIN

```python
%sql
-- UNOPTIMIZED (WE USED THE ORDER BY SO IT TOOK US AROUND 3S)
select 
c.first_name,c.last_name,od.product_name,od.price_each
from customers c  
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id
ORDER BY OD.price_each DESC;
```

```python
%sql
-- OPTIMIZED CODE (WE USED THE WHERE CLAUSE WITH THE ORDER BY SO IT TOOK US AROUND 2S)
-- IT IS NOT SHOWING MUCH DIFFERENCE IN THIS BECAUSE I HAVE A VERY SMALL DATASET, WHICH WON'T SHOW ANY EFFECT RIGHT NOW.
select 
c.first_name,c.last_name,od.product_name,od.price_each
from customers c  
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id
WHERE od.price_each > 1000
ORDER BY od.price_each DESC;
```

### GROUP BY WITHOUT FILTERING VS WITH FILTERING

```python
%sql
-- UNOPTIMIZED (WE USED THE GROUP BY WITHTOUT ANYFILTERING SO IT TOOK US AROUND 3S)
select 
c.customer_id,sum(od.price_each*od.quantity) as total_spent
from customers c  
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id
GROUP BY c.customer_id;
```

```python
%sql
-- OPTIMIZED QUERY (WE USED THE WHERE CLAUSE WITH THE GROUP BY SO IT TOOK US AROUND 3S)
-- IT IS NOT SHOWING MUCH DIFFERENCE IN THIS BECAUSE I HAVE A VERY SMALL DATASET, WHICH WON'T SHOW ANY EFFECT RIGHT NOW..
select 
c.customer_id,sum(od.price_each*od.quantity) as total_spent
from customers c  
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id
WHERE o.order_date >='2023-01-01'
GROUP BY c.customer_id;
```

### OBSERVING THE QUERY USING THE EXPLAIN CLAUSE/KEYWORD


```python
%sql
EXPLAIN
select 
c.first_name,c.last_name,od.product_name,od.price_each
from customers c  
join orders o on c.customer_id=o.customer_id
join order_details od on o.order_id=od.order_id
WHERE OD.price_each >1000;
```

### USING THE INDEX

```python
%sql
-- NORMAL QUERY WITHOUT INDEX (TOOK US 2S)
select * 
from order_details
where price_each > 1000;
```

```python
%sql
-- CREATING THE INDEX 
-- NOTE: Only tradional Relational databases like PostgreSQL, MySQL, or SQL Server.
-- Apache Spark SQL and Databricks do not support traditional B-Tree or column indexes via CREATE INDEX
CREATE INDEX idx_price on order_details(price_each);

-- RUNNNING THE QUERY 
select * 
from order_details
where price_each > 1000;
```
