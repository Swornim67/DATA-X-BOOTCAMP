```python
%sql
-- UNIQUE ROW NUMBER FOR EACH ROW.
select customer_id,order_id,order_date,
row_number() over (order by order_date) as row_number
from orders;
```

```python
%sql
-- SHOW THE FIRST PRODUCT BOUGHT BY THE CUSTOMER (purchase_sequence = 1 identifies the first product row for each custome)
select 
c.first_name,od.product_name,
row_number() over(partition by c.customer_id order by o.order_date) as purchase_sequence
from customers c  
join orders o 
on c.customer_id=o.customer_id
join order_details od 
on o.order_id=od.order_id
```

```python
%sql
-- RANK() - RANKS WITH THE GAP.
select 
product_name,quantity,
rank() over(order by quantity ) as ranking
from order_details 

-- RANK() assigns ranks to rows based on the quantity.
-- If multiple rows have the same quantity, they receive the same rank.
-- The next rank skips the number of positions occupied by tied rows,
-- creating a gap before continuing with the next rank.
```

```python
%sql
-- Identify the product with the highest purchased quantity in each order.
select 
order_id,product_name,quantity,
row_number() over(partition by order_id order by quantity desc) as BUYING_COUNT
from order_details 
```

```python
%sql
-- 6. Using the orders table, write a query to find only the earliest order for each customer. (USE CTE)
with customer_order as
(
select 
customer_id,order_id,order_date,
row_number() over(partition by customer_id order by order_date asc) as ranking
from orders 
)

select * from customer_order where ranking=1;
```
