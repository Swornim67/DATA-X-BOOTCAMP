### QUESTIONS


```python
%sql
-- 1. LIST ALL THE CUSTOMERS ORDERS WITH PRODUCT DETAILS 
select 
c.customer_id,c.first_name,c.last_name,od.product_name,od.quantity
from customers c 
join order_details od 
on c.customer_id=od.customer_id
```

```python
%sql
-- 2. FULL BREAKDOWN OF CUSTOMERS,ORDER AND THE ORDER DETAILS 
select 
c.customer_id,c.first_name,c.last_name,c.city,o.order_id,o.total_amount,od.product_id,od.product_name,od.quantity,od.price_each
from customers c 
join orders o 
on c.customer_id=o.customer_id
join order_details od 
on o.order_id=od.order_id
```

```python
%sql
-- 3. SHOW ALL THE CUSTOMERS WITH THEIR ORDER
select 
c.customer_id,c.first_name,c.last_name,o.order_id,o.total_amount
from customers c 
left join orders o
on c.customer_id=o.customer_id
```

```python
%sql
-- 4. SHOW ME THE CUSTOMERS WHO HAVEN'T ORDERED ANYTHING 
select 
c.customer_id,c.first_name,c.last_name,o.order_id
from customers c 
left join orders o
on c.customer_id=o.customer_id
where o.order_id is null
```

```python
%sql
-- 5. SHOW ALL THE CUSTOMERS WITH THEIR ORDER (RIGHT JOIN)
select 
c.customer_id,c.first_name,c.last_name,o.order_id,o.total_amount
from orders  o 
right join customers c 
on o.customer_id=c.customer_id
```

```python
%sql
-- FOR THE SELF JOIN WE ARE USING THE TABLE EMP
select * from emp;
```

```python
%sql
-- 6. SHOW ME THE LIST OF EMPLOYEES WITH THEIR MAMAGER NAME
select 
e.emp_id,e.first_name,e.last_name,m.first_name as manager_name
from emp e
left join emp m
on e.manager_id=m.emp_id
```

```python
%sql
-- 7. FIND THE TOTAL NO OF PEOPLE REPORTING TO THEIR MANAGERS (EMPLOYEE COUNT OF EACH MANAGERS)
select 
m.first_name,count(e.emp_id) as employee_count
from emp e
join emp m 
on e.manager_id=m.emp_id
group by m.first_name;
```

```python
%sql
-- 8. FULL AUDIT OF THE CUSTOMERS AND THEIR ORDERS (ORDERED + NOT ORDERED PEOPLES)
select 
c.first_name,o.order_id,od.product_name
from customers c 
full outer join orders o  
on c.customer_id=o.customer_id
full outer join order_details od
on o.order_id=od.order_id;
```

```python
%sql
-- 9. GET THE LIST OF UNIQUE CUSTOMERS CITIES + UNIQUE PRODUCT CATEGORIES
select distinct city from customers 
union 
select distinct category from order_details ;
```

```python
%sql
-- 10. GET THE LIST OF ALL CUSTOMERS CITIES + PRODUCT CATEGORIES (INCLUDING THE DUPLICATES)
select  city from customers 
union all 
select category from order_details;
```

```python
%sql
-- 11. TOP 3 CUSTOMERS BY SPEND
select 
c.customer_id,c.first_name,c.last_name,sum(o.total_amount * od.quantity) as total_spent
from customers c 
join orders o 
on c.customer_id=o.customer_id
join order_details od 
on o.order_id=od.order_id
group by c.customer_id,c.first_name,c.last_name
order by total_spent desc limit 3;
```
