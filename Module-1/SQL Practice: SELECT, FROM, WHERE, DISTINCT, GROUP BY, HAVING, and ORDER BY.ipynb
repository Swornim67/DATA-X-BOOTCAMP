**SQL Practice: SELECT, FROM, WHERE, DISTINCT, GROUP BY, HAVING, and ORDER BY**

```sql
-- Checking the data 
select * from employees;
```

```sql
-- show me the emp,id,first name and salary from employees
select emp_id,first_name,salary from employees;
```

```sql
-- Showing only those people who's salary is greater then 80k  
select first_name,salary 
from employees 
where salary > 80000;
```

```sql
-- Showing only those people who's salary is greater then 100k
select first_name,salary from employees
where salary > 100000;
```

```sql
-- show me the inactive people details
select * from employees where status='Inactive';
```

```sql
-- Show me the active people and living in city mumbai
select emp_id,first_name,status,city from employees 
where status='Active' and city='Mumbai';
```

```sql
-- Show me the details of the people who are working in department Marketing and Finance
select emp_id,first_name,department from employees 
where department in('Marketing','Finance');
```

```sql
-- Show me the details of the people who are working either in department marketing or finance
select emp_id,first_name,department from employees 
where department ='Marketing' or department='Finance';
```

```sql
-- Show me the list of people who are working in department marketing with salary greater then 80k
select * from employees 
where department='Marketing' and salary > 80000;
```

```sql
-- Show me the list of people who joined the company before 2020
select * from employees 
where hire_date < ' 2020-01-01'
```

```sql
-- Show me the list of people who joined the company after 2020
select * from employees 
where hire_date > '2020-01-01'
```

```sql
-- Show me the list of person from the highest to lowest salary
select * from employees 
order by salary desc ;
```

```sql
-- Show me alphabetically first name and department 
select * from employees 
order by first_name ,department ;
```

```sql
-- Show me alphabetically ascending wise department and descending wise first_name
select * from employees 
order by department asc, first_name desc
```

```sql
-- Show me the list of recently hired person 
select first_name,hire_date from employees 
order by hire_date desc;
```

```sql
-- show me the highest salary person details 
select * from employees 
order by salary desc limit 1;
```

```sql
-- show me the 2nd highest salary person detail 
select * from employees 
order by salary desc 
limit 1 offset 1;
```

```sql
-- show me the recently 2nd hired person detail 
select * from employees 
order by hire_date desc limit 1 offset 1;
```

```sql
-- remove the duplicate first_name from the list 
select distinct first_name from employees;
```

```sql
-- remove the duplicate salary numbers from the list 
select distinct salary from employees;
```

```sql
-- remove the duplicate city from the list
select distinct city from employees;
```

```sql
-- Year wise hiring report list 
select year(hire_date) as hire_year,
count(emp_id) as total_hired
from employees 
group by year(hire_date)
order by hire_year asc;
 
```

```sql
-- Show me the anaotomy of the table
describe employees;
```

```sql
-- Count me the no of rows in the table
select count(*) from employees;
```

```sql
select count(1) from employees; -- It counts the total number of rows in the employees table and returns a single number.
```

```sql
select 1 from employees;    -- prints the number 1 for every row present in the employees table
```

```sql
-- show me the details of the first row
select * from employees 
limit 1;
```

```sql
-- count me the no of null columns from the email 
select count(*) from employees 
where email is null;
```

```sql
-- show me the null values presented in the email 
select * from employees 
where email is null;
```

```sql
-- show me the name of peoples who's email isn't null
select * from employees 
where email is not null;
```

```sql
-- show me the list of department 
select distinct department from employees;
```

```sql
-- count me the no of people in the department 
select department,count(*)as `People working in the department`from employees 
group by department;
```

```sql
-- show me the department where people are less then 9 working in it 
select department,count(*) from employees 

group by department
having count(*) <9;
```
