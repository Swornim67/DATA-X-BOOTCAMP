```python
%sql
-- LIST ALL THE CUSTOMERS NAME + THEIR TOTAL SPEND + SPENDING CATEGORY 
with customer_spent as (
select c.customer_id,concat(first_name,' ',last_name) as full_name,
coalesce(sum(od.quantity*od.price_each),0) as total_spent
from customers c
left join orders o 
on c.customer_id=o.customer_id
left join order_details od
on o.order_id=od.order_id
group by c.customer_id,c.first_name,c.last_name
)

select full_name,total_spent,  
    case 
        when total_spent = 0 then 'No orders'
        when total_spent < 500 then 'Bronze'
        when total_spent between 1000 and 2000 then 'Silver'
        else 'Gold'
        end as customer_category
from customer_spent;
```

```python
%sql
-- Q8. Display each customer's full name, total spending, and spending category
-- Categories:
-- Zero spending → No Orders
-- Above 0 but below 500 → Bronze
-- 1,000 through 2,000 → Silver
-- All other amounts → Gold

with customer_details as 
(
    select concat(c.first_name,' ',c.last_name) as full_name,
    coalesce(sum(o.total_amount),0) as total_spent,
        case 
            when coalesce(sum(o.total_amount),0) = 0 then 'No Orders'
            when sum(o.total_amount) < 500 then 'Bronze'
            when sum(o.total_amount) between 1000 and 2000 then 'Silver'
            else 'Gold'
        end as category
    from customers c 
    left join orders o
    on c.customer_id=o.customer_id
    group by c.customer_id,c.first_name,c.last_name    
)
select full_name,total_spent,category from customer_details
```
