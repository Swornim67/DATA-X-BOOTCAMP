```python
%sql
-- FIND THE CUSTOMERS WHO HAVE PLACED THE ORDER ABOVE THE AVG ORDER AMOUNT.
select distinct c.first_name,c.last_name,o.total_amount
from customers c 
join orders o 
on c.customer_id=o.customer_id
where o.total_amount > 
(select avg(total_amount) from orders );
```

```python
%sql
-- LIST ALL PRODUCTS THAT HAVE BEEN SOLD MORE THEN 2 TIMES IN TOTAL.
select od.product_name,sum(od.quantity) as total_sold
from order_details od
group by od.product_name
having sum(od.quantity) > 
(select 2)
```

```python
%sql
-- 1. Find orders whose total amount is greater than the average order amount
select order_id,customer_id,total_amount 
from orders 
where total_amount >
(select avg(total_amount)from orders )
```

```python
%sql
-- 2. Find customers who have placed at least one order.
select customer_id,first_name,last_name 
from customers 
where customer_id in 
(select customer_id from orders )
```

```python
%sql
-- 3. Find products whose total quantity sold is greater than the average quantity sold per product.
select product_name,sum(quantity) as total_sold 
from order_details 
group by product_name
having sum(quantity) >
(select avg(avg_product)
from 
(select product_name,sum(quantity)as avg_product
from order_details 
group by product_name) as product_sales
);
```

```python
%sql
-- 4. Find customers who have never placed an order.
select customer_id,first_name,last_name 
from customers 
where customer_id not in 
(select customer_id from orders);
```

```python
%sql
-- 5. Find products whose total quantity sold is greater than 5
select product_name,total_quantity
from 
(select product_name,sum(quantity) as total_quantity
from order_details 
group by product_name) as quantity_count
where total_quantity > 5;
```
