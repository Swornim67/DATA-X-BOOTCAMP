```python
%sql
-- 2. Using your orders table, write a query to show each customer's order date alongside the previous order date for that same customer, in chronological order.
select 
order_id,customer_id,order_date,
lag(order_date) over(partition by customer_id order by order_date) as previous_date 
from orders 
```

```python
%sql
-- 4. Using your order_details table, write a query to calculate the cumulative quantity purchased, separately for each customer, in order of purchase date.
select
customer_id,order_id,product_name,quantity,
sum(quantity) over (partition by customer_id order by order_id,order_detail_id,order_date ) as cumulative_quantity
from order_details;
```
